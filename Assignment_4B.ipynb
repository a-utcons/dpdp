### Problem Statement 4B

4B. Implement the Continuous Bag of Words (CBOW) Model for the twitter dataset (download from Kaggle) using the below steps:
a. Data Collection and Preprocessing: Use Kaggle Twitter dataset or any dataset with tweets. Tokenize the tweets, clean the text, and prepare it for training.
b. Train Word2Vec CBOW Model
c. Evaluate Results: After training, we'll check the learned word embeddings and possibly evaluate using some examples.

```
# Optional installation:
# !pip install gensim

import os
import pandas as pd
from gensim.models import Word2Vec
```

```
# Step a: Data Collection and Preprocessing
csv_path = "../LP IV Updated Datasets/CBOW/twitter_dataset.csv"

data = pd.read_csv(csv_path)
print("Data shape:", data.shape)
print(data.head(3))

# Tokenize tweets and clean text into lowercase words
tweet_sentences = []
for text in data['Text']:
    if isinstance(text, str):
        words = [w.lower() for w in text.split() if w.isalpha() and len(w) > 2]
        if len(words) > 1:
            tweet_sentences.append(words)

print(f"Total prepared tweet sentences: {len(tweet_sentences)}")
print("Sample cleaned tweet:", tweet_sentences[0])
```

```
# Step b: Train Word2Vec CBOW Model
# sg=0 specifies the Continuous Bag of Words (CBOW) architecture
cbow_model = Word2Vec(
    sentences=tweet_sentences,
    vector_size=100,
    window=5,
    min_count=2,
    sg=0,
    epochs=10
)

print("Vocabulary size learned:", len(cbow_model.wv))
```

```
# Step c: Evaluate Results - check learned word embeddings with examples
# Find words with similar context
sample_words = ['good', 'happy', 'day', 'time', 'work']

for word in sample_words:
    if word in cbow_model.wv:
        print(f"\nWords most similar to '{word}':")
        for similar_word, similarity_score in cbow_model.wv.most_similar(word, topn=3):
            print(f"  -> {similar_word}: {similarity_score:.4f}")

# Check similarity score between two words
if 'good' in cbow_model.wv and 'happy' in cbow_model.wv:
    sim = cbow_model.wv.similarity('good', 'happy')
    print(f"\nCosine similarity between 'good' and 'happy': {sim:.4f}")
```
