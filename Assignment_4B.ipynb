### Problem Statement 4B

Implement the Continuous Bag of Words (CBOW) Model for the twitter dataset (download from Kaggle) using the below steps:
a. Data Collection and Preprocessing: Use Kaggle Twitter dataset or any dataset with tweets. Tokenize the tweets, clean the text, and prepare it for training.
b. Train Word2Vec CBOW Model
c. Evaluate Results: After training, we'll check the learned word embeddings and possibly evaluate using some examples.

```
# Step 1: Import libraries
import re
import pandas as pd
from gensim.models import Word2Vec
```

```
# Step 2: Load dataset
df = pd.read_csv("../LP IV Updated Datasets/CBOW/twitter_dataset.csv")

print(df.head())
```

```
# Step 3: Clean and tokenize tweets
def clean_tweet(text):

    if not isinstance(text, str):
        return []

    text = re.sub(r'http\S+|www\S+', '', text)
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'#\w+', '', text)
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    words = text.lower().split()

    return [w for w in words if len(w) > 2]


df['clean_tokens'] = df['Text'].apply(clean_tweet)

tweet_sentences = [
    tokens for tokens in df['clean_tokens']
    if len(tokens) > 1
]

print("Total tweets:", len(tweet_sentences))
print("Sample:", tweet_sentences[0])
```

```
# Step 4: Train CBOW model
cbow_model = Word2Vec(
    sentences=tweet_sentences,
    vector_size=100,
    window=5,
    min_count=2,
    sg=0,
    epochs=10
)

print("Vocabulary size:", len(cbow_model.wv))
```

```
# Step 5: Find similar words
sample_words = ['good', 'happy', 'day', 'time', 'work']

for word in sample_words:
    if word in cbow_model.wv:
        print("\nSimilar words to:", word)

        for similar_word, score in cbow_model.wv.most_similar(word, topn=3):
            print(similar_word, score)
```

```
# Step 6: Calculate similarity
if 'good' in cbow_model.wv and 'happy' in cbow_model.wv:
    sim = cbow_model.wv.similarity('good', 'happy')
    print("Cosine similarity:", sim)
```
