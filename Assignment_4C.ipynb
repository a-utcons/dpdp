### Problem Statement 4C

4C. Implement the Continuous Bag of Words (CBOW) Model for the given textual document 3 using the below steps:
a. Data preparation
b. Train Word2Vec CBOW Model
c. Evaluate Results: possibly evaluate using some examples.

```
# Optional installation:
# !pip install gensim

import os
from gensim.models import Word2Vec
```

```
# Step a: Data preparation
doc_path = "../LP IV Updated Datasets/CBOW/document3.txt"

with open(doc_path, 'r', encoding='utf-8') as f:
    text = f.read()

print("Document preview:\n", text[:200])

# Split text into sentences and words
sentences = []
for line in text.split('.'):
    words = [w.lower() for w in line.split() if w.isalpha()]
    if len(words) > 1:
        sentences.append(words)

print(f"\nPrepared {len(sentences)} sentences for CBOW training.")
print("Sample sentence words:", sentences[0])
```

```
# Step b: Train Word2Vec CBOW Model
# sg=0 specifies CBOW architecture
cbow_doc = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    sg=0,
    epochs=50
)

print("Vocabulary size learned from document 3:", len(cbow_doc.wv))
print("Learned words:", list(cbow_doc.wv.index_to_key))
```

```
# Step c: Evaluate Results
# Evaluate semantic similarity for key words in document 3
test_words = ['virus', 'influenza', 'transmission']

for word in test_words:
    if word in cbow_doc.wv:
        print(f"\nWords most similar to '{word}':")
        for sim_word, score in cbow_doc.wv.most_similar(word, topn=3):
            print(f"  -> {sim_word}: {score:.4f}")
```
