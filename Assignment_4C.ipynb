### Problem Statement 4C

4C. Implement the Continuous Bag of Words (CBOW) Model for the given textual document 3 using the below steps:
a. Data preparation
b. Train Word2Vec CBOW Model
c. Evaluate Results: possibly evaluate using some examples.

```
# Step 1: Import libraries
from gensim.models import Word2Vec
```

```
# Step 2: Load document
doc_path = "../LP IV Updated Datasets/CBOW/document3.txt"

with open(doc_path, 'r', encoding='utf-8') as f:
    text = f.read()

print("Document preview:", text[:200])
```

```
# Step 3: Split into sentences and words
sentences = []

for line in text.split('.'):
    words = [
        w.lower() for w in line.split()
        if w.isalpha()
    ]

    if len(words) > 1:
        sentences.append(words)

print("Total sentences:", len(sentences))
print("Sample:", sentences[0])
```

```
# Step 4: Train CBOW model
cbow_doc = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    sg=0,
    epochs=50
)
```

```
# Step 5: Check vocabulary
print("Vocabulary size:", len(cbow_doc.wv))
print("Learned words:", list(cbow_doc.wv.index_to_key))
```

```
# Step 6: Find similar words
test_words = ['virus', 'influenza', 'transmission']

for word in test_words:

    if word in cbow_doc.wv:

        print("\nSimilar words to:", word)

        for sim_word, score in cbow_doc.wv.most_similar(word, topn=3):
            print(sim_word, score)
```
